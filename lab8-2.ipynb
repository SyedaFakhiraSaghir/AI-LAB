```python
# Task 1:
# Consider a standard 6-sided die. Calculate the probability of getting an even number when
# rolling the die. Calculate the probability of rolling a number greater than 4, and then the
# probability of rolling a number less than 3. Consider two dice. What is the probability of the
# sum of the numbers being 7 or more? What is the probability of the sum being exactly 8?
# Given that the first die shows a number greater than 4, what is the probability that the
# second die shows an odd number?

even= [2, 4, 6]
total_outcomes = [1, 2, 3, 4, 5, 6]
print(f"Probability of an even number: {len([2, 4, 6]) / len(total_outcomes)}")
print(f"Probability of number greater than 4: { len([5,6]) / len(total_outcomes)}")
print(f"Probability of number less than 3: {len([1, 2]) / len(total_outcomes)}")

sumGT7 = 0
for die1 in range(1, 7):
    for die2 in range(1, 7):
        if die1 + die2 >= 7:
            sumGT7 += 1
print(f"Probability of the sum being 7 or more: {sumGT7 / 36}")

sum8 = 0
for die1 in range(1, 7):
    for die2 in range(1, 7):
        if die1 + die2 == 8:
            sum8 += 1
print(f"Probability of the sum being exactly 8: {sum8 / 36}")

d2odd = 0
d1GT4 = 0
for die1 in range(1, 7):
    for die2 in range(1, 7):
        if die1 > 4:
            d1GT4 += 1
            if die2 % 2 != 0:
                d2odd += 1
print(f"Probability of the second die being odd given the first is greater than 4: {d2odd / d1GT4}")
```

```python
# Task 2:
# Consider a standard deck of 52 cards. What is the probability of drawing a red card (hearts
# or diamonds)? Now, given that you have drawn a red card, what is the probability that the
# card is a heart? Given that the drawn card is a face card (Jack, Queen, King), what is the
# probability that it is a diamond? Given that the card drawn is a face card, calculate the
# probability that it is also a spade or a queen.

suits = ["Hearts", "Diamonds", "Clubs", "Spades"]
ranks = ["2", "3", "4", "5", "6", "7", "8", "9", "10", "Jack", "Queen", "King", "Ace"]

deck = [(rank, suit) for rank in ranks for suit in suits]

red_cards = [card for card in deck if card[1] in ["Hearts", "Diamonds"]]
print(f"Probability of drawing a red card: {len(red_cards) / len(deck)}")

hearts = [card for card in deck if card[1] == "Hearts"]
print(f"Probability of drawing a Heart given it's a red card: {len(hearts) / len(red_cards)}")

face_cards = [card for card in deck if card[0] in ["Jack", "Queen", "King"]]
diamond_face_cards = [card for card in face_cards if card[1] == "Diamonds"]
print(f"Probability of drawing a Diamond given it's a face card: {len(diamond_face_cards) / len(face_cards)}")

spade_or_queen_face_cards = [card for card in face_cards if card[1] == "Spades" or card[0] == "Queen"]
print(f"Probability of drawing a Spade or a Queen given it's a face card: {len(spade_or_queen_face_cards) / len(face_cards)}")
```

```python
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination

model = DiscreteBayesianNetwork([('Disease', 'Fever'),
                                 ('Disease', 'Cough'),
                                 ('Disease', 'Fatigue')])

cpd_disease = TabularCPD(variable='Disease', variable_card=2,
                         values=[[0.9], [0.1]],
                         state_names={'Disease': ['NoFlu', 'Flu']})

cpd_fever = TabularCPD(variable='Fever', variable_card=2,
                       values=[[0.9, 0.2],
                               [0.1, 0.8]],
                       evidence=['Disease'], evidence_card=[2],
                       state_names={'Fever': ['No', 'Yes'], 'Disease': ['NoFlu', 'Flu']})

cpd_cough = TabularCPD(variable='Cough', variable_card=2,
                       values=[[0.8, 0.3],
                               [0.2, 0.7]],
                       evidence=['Disease'], evidence_card=[2],
                       state_names={'Cough': ['No', 'Yes'], 'Disease': ['NoFlu', 'Flu']})

cpd_fatigue = TabularCPD(variable='Fatigue', variable_card=2,
                         values=[[0.7, 0.4],
                                 [0.3, 0.6]],
                         evidence=['Disease'], evidence_card=[2],
                         state_names={'Fatigue': ['No', 'Yes'], 'Disease': ['NoFlu', 'Flu']})

model.add_cpds(cpd_disease, cpd_fever, cpd_cough, cpd_fatigue)
assert model.check_model()

infer = VariableElimination(model)

posterior_prob = infer.query(variables=['Disease'],
                             evidence={'Fever': 'Yes', 'Cough': 'Yes'})
print("Probability of Disease given Fever and Cough:\n", posterior_prob)

model.add_node('Chills')
model.add_edge('Disease', 'Chills')

cpd_chills = TabularCPD(variable='Chills', variable_card=2,
                        values=[[0.8, 0.25],
                                [0.2, 0.75]],
                        evidence=['Disease'], evidence_card=[2],
                        state_names={'Chills': ['No', 'Yes'], 'Disease': ['NoFlu', 'Flu']})

model.add_cpds(cpd_chills)
assert model.check_model()

infer = VariableElimination(model)

posterior_prob_with_chills = infer.query(variables=['Disease'],
                                         evidence={'Fever': 'Yes', 'Cough': 'Yes', 'Chills': 'Yes'})
print("\nProbability of Disease given Fever, Cough, and Chills:\n", posterior_prob_with_chills)

posterior_prob_fatigue = infer.query(variables=['Fatigue'], evidence={'Disease': 'Flu'})
print("\nProbability of Fatigue given Disease is Flu:\n", posterior_prob_fatigue)
```

```python
# Task 4:
# Define a simple Markov Model with three weather states: Sunny, Cloudy, and Rainy. Create
# a transition matrix for the weather states, specifying the probabilities of transitioning from
# one state to another (e.g., from Sunny to Cloudy, or from Cloudy to Rainy). Simulate the
# weather for the next 10 days, starting with a Sunny day, using the Markov Model. Calculate
# the probability of having at least 3 rainy days over the 10-day period.
# Install pgmpy if not already installed
# pip install pgmpy

import numpy as np
import random
weather_states = ['Sunny', 'Cloudy', 'Rainy']
transition_matrix = np.array([
    [0.6, 0.3, 0.1],
    [0.3, 0.4, 0.3],
    [0.2, 0.3, 0.5]
])
def weather_forecast(days, initial_state):
    weather_sequence = [initial_state]
    current_state = initial_state
    for _ in range(days -1 ):
        change_probs = transition_matrix[weather_states.index(current_state), :]
        next_state = random.choices(weather_states, change_probs)[0]
        weather_sequence.append(next_state)
        current_state = next_state
    return weather_sequence

weather = weather_forecast(10, 'Sunny')
print("Weather forecast [next 10 days]:", weather)
num_simulations = 10000
rainy_days_counts = []
for _ in range(num_simulations):
    weather_sim = weather_forecast(10, 'Sunny')
    rainy_days_counts.append(weather_sim.count('Rainy'))

prob_at_least_3_rainy_days = sum(1 for count in rainy_days_counts if count >= 3) / num_simulations
print(f"Probability of 3 rainy days in 10 days: {prob_at_least_3_rainy_days}")
```
